# Phase 4 – Garbage Detection EDA Notebook

This notebook analyzes image and label counts across train, validation, and test splits and inspects `datasets/garbage/data.yaml`.

```python
import os
import yaml

dataset = "datasets/garbage"
if not os.path.exists(dataset):
    dataset = "../datasets/garbage"

print("Train Images:", len(os.listdir(f"{dataset}/train/images")))
print("Train Labels:", len(os.listdir(f"{dataset}/train/labels")))

print("Validation Images:", len(os.listdir(f"{dataset}/valid/images")))
print("Validation Labels:", len(os.listdir(f"{dataset}/valid/labels")))

print("Test Images:", len(os.listdir(f"{dataset}/test/images")))
print("Test Labels:", len(os.listdir(f"{dataset}/test/labels")))

with open(f"{dataset}/data.yaml") as f:
    config = yaml.safe_load(f)

print("\n--- YAML Configuration ---")
print(config)
print("Classes count (nc):", config.get("nc"))
print("Class names:", config.get("names"))
```
